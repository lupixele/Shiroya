```python
import pandas as pd
import numpy as np
data = {
    'Name': ['Asha', np.nan, 'Kiran', 'Ravi'],
    'City': ['Hyderabad', 'Chennai', np.nan, 'Delhi']
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
df.fillna("Unknown", inplace=True)
print("\nDataFrame after filling NaN values:")
print(df)
```

```python
import pandas as pd
data = {
    'Name': ['Asha', 'Ravi', 'Kiran', 'Priya'],
    'Marks': [85, 92, 78, 88]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
sorted_df = df.sort_values(by='Marks')
print("\nDataFrame sorted by Marks (Ascending):")
print(sorted_df)
```

```python
import pandas as pd
data = {
    'Department': ['CSE', 'CSE', 'ECE', 'ECE', 'IT'],
    'Name': ['Asha', 'Ravi', 'Kiran', 'Priya', 'Rahul'],
    'Marks': [85, 90, 78, 88, 95]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
grouped = df.groupby('Department')['Marks'].mean()
print("\nAverage Marks by Department:")
print(grouped)
```
