```python
import pandas as pd
# Text file
text_data = pd.read_csv("students.txt")
print("Text File:")
print(text_data)

# CSV file
csv_data = pd.read_csv("students.csv")
print("\nCSV File:")
print(csv_data)

# Excel file
excel_data = pd.read_excel("students.xlsx")
print("\nExcel File:")
print(excel_data)

# JSON file
json_data = pd.read_json("students.json")
print("\nJSON File:")
print(json_data)
```
